# Работа с таблицами

## Как работать с таблицей при помощи инструмента, который не знает, что такое таблица?

Таблица - это вид представления данных в виде строк и столбцов. При этом у столбцов бывают названия, а у строк - номера. Получается, что каждая строка - это `вхождение` данных, которые представлены однообразно. Например, в [этой таблице](https://docs.google.com/spreadsheets/d/1nUysSVuCLI0alRInTMzNnCk1EB2-wRMS1ssjElUb2EY/edit?usp=sharing) каждая строка содержит информацию по одной работе, и для каждой работы есть ее номер, название, город, секция, формулировка грамоты и вид диплома. Все эти данные подписаны в первой строке таблицы - строке-заголовке.

"Подписи" существуют в типе данных, который мы недавно прошли: это словари. Ключи словаря подписывают данные так же, как названия столбцов таблицы. Поэтому каждую строку таблицы можно представить в виде словаря, в котором ключи - названия столбцов, а значения - данные из ячейки этой строки, соответствующей каждому из названий столбцов. Получается, мы можем составить словарь для отображения каждой строки таблицы. А теперь мы можем эти словари объединить в список.

Давайте попробуем прочитать таблицу выше и вывести данные по первым трем работам.

Для работы нам понадобится модуль excel. Его нужно установить через терминал, если он не установлен (`pip install excel`) и импортировать командой `import excel`. После этого мы сможем вызывать методы этого модуля.

Таблицу сначала надо скачать, а затем считать в формат JSON. Обращайте внимание, куда вы сохраняете таблицу и куда - ваш код.

А теперь выведите информацию по работе, располагающейся в 115-й строке таблицы.

А теперь получите название работы с номером 241273.
