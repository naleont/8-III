```python
organisation = df.groupby('Организация').agg({
    'Номер работы': 'count', 
    'Лауреатство': lambda x: (x == 'Да').sum()
}).rename(columns={
    'Номер работы': 'Всего работ', 
    'Лауреатство': 'Лауреатов'
})
```

```python
print(organisation.sort_values('Лауреатов', ascending=False))
```
