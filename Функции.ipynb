Мы помним, что чем короче код, тем лучше. Часто бывает так, что нам приходится писать в разных местах очень похожий код. Особо умные среды разработки даже подсвечивают это как ошибку.

Почему нам нужен этот похожий код? Например, часто нам нужно выполнить какое-то действие несколько раз с разными данными.

А еще реальные программы могут иметь сложную структуру. Для того, чтобы программу было легче читать, удобно записать ее как линейную последовательность действий и отдельно записать, из чего каждое из этих действий состоит.

Для решения этих задач на помощь приходят функции. Функцию можно рассматривать как отдельную подпрограмму в коде одной общей программы. Каждая функция имеет название и параметры. Параметры - это то же, что аргументы, только аргументами мы их называем, когда применяем функцию, а параметрами - когда ее создаем и описываем.

Для создания функции используется ключевое слово `def`, после него пишется название функции, скобки с параметрами (скобки должны быть даже если функция не будет принимать аргументы) и двоеточие.

Код функции пишется вложенным в функцию. То есть все, что принадлежит функции, должно быть записано с отступом относительно ключевого слова `def`.

Если функция должна в качестве результата своей работы выдать какое-то значение, пользуются оператором `return`.

```python
# Пример: описываем функцией решение уравнения вида ax + b = 0

def equation(a, b):
  if a == 0:
    if b == 0:
        x = "INF"
    else:
        x = "NO"
  else:
      x = -b / a
  return x
```

При запуске этого кода кажется, что

```python
# Вызываем функцию с аргументами 3 и 9

solution_1 = equation(3, 9)
print(solution_1)
```

```python
# А теперь вызываем функцию с другими аргументами.
# Код функции уже написан, поэтому нам не нужно писать его заново: достаточно заново вызвать функцию.

solution_2 = equation(15, 3)
print(solution_2)
```

```python
today = 'wednesday'

if today == 'monday':
  solution = equation(15, 3)
else:
  solution = equation(16, 2)

print(solution)
```

Вспомним страшную задачу про билеты на метро. В ней билет на 1 поездку стоит 15 рублей, на 10 - 125 рублей, на 60 - 440 рублей, и надо посчитать, какой набор билетов будет наиболее выгодным для совершения определенного количества поездок.

Давайте сделаем задачу длиннее: будем спрашивать, сколько поездок хочет совершить пассажир, а потом рассчитывать, сколько денег он потратит на это количество поездок.

```python
# Сначала задаем переменными цены на билеты

price_1 = 15
price_10 = 125
price_60 = 440

# Описываем функцию, рассчитывающую наиболее дешевый набор билетов
# Функция принимает количество поездок, а возвращает 3 значения: количество билетов на 1, 10 и 60 поездок
def cheapest_rides(rides):
  tickets_60 = rides // 60
  remainder_60 = rides % 60

  tickets_10 = remainder_60 // 10
  tickets_1 = remainder_60 % 10

  if tickets_1 * price_1 + tickets_10 * price_10 > price_60:
      tickets_60 += 1
      tickets_10 = 0
      tickets_1 = 0

  if tickets_1 * price_1 > price_10:
      tickets_10 += 1
      tickets_1 = 0

  return tickets_1, tickets_10, tickets_60

# Описываем функцию, рассчитывающую общую стоимость поездок
# Функция принимает количество билетов на 1, 10 и 60 поездок и возвращает общую стоимость
def calculate_cost(tickets_1, tickets_10, tickets_60):
  total_cost = tickets_1 * price_1 + tickets_10 * price_10 + tickets_60 * price_60
  return total_cost

# Получаем от пользователя количество поездок
rides_count = int(input('Сколько поездок вы хотите совершить? '))

# С полученным значением вызываем функцию cheapest_rides
t_1, t_2, t_3 = cheapest_rides(rides_count)

# Со значениями, полученными из предыдущей функции, вызываем функцию calculate_cost
cost = calculate_cost(t_1, t_2, t_3)

# Печатаем получившееся значение
print(cost)
```

Посмотрим, что у нас получилось: программа состоит из задания трех переменных, двух функций, а основной код занимает всего 4 строки. Функции можно будет переиспользовать несколько раз с теми же или другими значениями, а код стал более человекочитаемым.
